```python
#Graficar los números pares utilizando un ciclo
import matplotlib.pyplot as plt

numeros_pares = []

for numero in range(2, 21, 2):
    numeros_pares.append(numero)

print("numeros pares:", numeros_pares)

# Graficar los números pares
plt.bar(range(len(numeros_pares)), numeros_pares, tick_label=numeros_pares)
plt.title("numeros pares del 2 al 20")
plt.xlabel("posicion")
plt.ylabel("valor")
plt.show()
```

```python
#Graficar temperaturas y clasificarlas mediante condicionales
import matplotlib.pyplot as plt
dias=["lunes","martes","miercoles","jueves","viernes"]
temperaturas=[18,22,31,27,16]

for i in range(len(temperaturas)):
    if temperaturas[i] >= 30:
        clasificacion = "caluroso"
    elif temperaturas[i] >= 20:
        clasificacion = "templado"
    else:
        clasificacion = "frio"
    print(dias[i], temperaturas[i], "°C", clasificacion)

plt.bar(dias,temperaturas)
plt.title("temperaturas de las semanas")
plt.xlabel("dia")
plt.ylabel("temperatura(°C)")
plt.show()
```

```python
#. Graficar las notas aprobadas y reprobadas
import matplotlib.pyplot as plt
calificaciones=[85,62,70,55,91,68,77]
aprobados=0
reprobados=0

for nota in calificaciones:
    if nota>=70:
        aprobados+=1
    else:
        reprobados+=1

print("aprobados", aprobados)
print("reprobados", reprobados)

plt.bar(["aprobados","reprobados"],[aprobados, reprobados])
plt.title("cantidad de aprobados y reprobados")
plt.xlabel("resultado")
plt.ylabel("cantidad")
plt.show()
```

```python
#Graficar una tabla de multiplicar
import matplotlib.pyplot as plt
numero=int(input("ingrese numero: "))
multiplicadores= list(range(1,11))
productos=[]
for multiplicador in multiplicadores:
  producto=numero*multiplicador
  productos.append(producto)
  print(numero, "x", multiplicador, "=", producto)

plt.plot(multiplicadores,productos,marker="o")
plt.title("tabla de multiplicar")
plt.xlabel("multiplicador")
plt.ylabel("producto")
plt.xticks(multiplicadores)
plt.grid(True)
plt.show()
```

```python
#Clasificar las calificaciones de los estudiantes

import matplotlib.pyplot as plt

notas = []
clasificaciones = []

for i in range(5):
    nota = float(input(f"Ingrese la nota del estudiante {i + 1}: "))
    notas.append(nota)

    if nota >= 90:
        clasificacion = "Excelente"
    elif nota >= 70:
        clasificacion = "Aprobado"
    else:
        clasificacion = "Reprobado"

    clasificaciones.append(clasificacion)

for i in range(5):
    print(f"Estudiante {i + 1}: {notas[i]} - {clasificaciones[i]}")

plt.bar(
    ["Est. 1", "Est. 2", "Est. 3", "Est. 4", "Est. 5"],
    notas,
    color="skyblue"
)
plt.axhline(90, color="green", linestyle="--", label="Excelente (90)")
plt.axhline(70, color="red", linestyle="--", label="Aprobado (70)")
plt.title("Calificaciones de los estudiantes")
plt.xlabel("Estudiantes")
plt.ylabel("Nota")
plt.legend()
plt.show()
```

```python
#Clasificar las temperaturas de una semana

import matplotlib.pyplot as plt

temperaturas = []
dias = ["Lunes", "Martes", "Miércoles", "Jueves",
        "Viernes", "Sábado", "Domingo"]

for i in range(7):
    temperatura = float(
        input(f"Ingrese la temperatura del {dias[i]}: ")
    )
    temperaturas.append(temperatura)

    if temperatura >= 30:
        clasificacion = "Calurosa"
    elif temperatura >= 20:
        clasificacion = "Agradable"
    else:
        clasificacion = "Fría"

    print(f"{dias[i]}: {temperatura} °C - {clasificacion}")

plt.plot(dias, temperaturas, marker="o", color="orange")
plt.title("Temperaturas de la semana")
plt.xlabel("Días")
plt.ylabel("Temperatura (°C)")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

ventas = []
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]

for i in range(5):
    venta = float(
        input(f"Ingrese las ventas del {dias[i]} en colones: ")
    )
    ventas.append(venta)

    if venta >= 100000:
        clasificacion = "Venta alta"
    elif venta >= 50000:
        clasificacion = "Venta media"
    else:
        clasificacion = "Venta baja"

    print(f"{dias[i]}: ₡{venta:,.2f} - {clasificacion}")

plt.bar(dias, ventas, color="mediumseagreen")
plt.title("Ventas de la tienda durante cinco días")
plt.xlabel("Días")
plt.ylabel("Ventas (colones)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
