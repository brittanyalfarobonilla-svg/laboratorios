```python
#1
ph = float(input("Ingrese un pH entre 0 y 14: "))

while ph < 0 or ph > 14:
    print("Valor inválido")
    ph = float(input("Ingrese un pH entre 0 y 14: "))

print("pH válido:", ph)
```

```python
#2
cantidad = int(input("Ingrese la cantidad de átomos: "))

contador = 1

while contador <= cantidad:
    print(contador)
    contador += 1
```

```python
#3
masa_total = 0

while masa_total < 100:
    masa = float(input("Ingrese una masa en gramos: "))
    masa_total = masa_total + masa
    print("Masa acumulada:", masa_total)

print("Se alcanzaron al menos 100 gramos")
```

```python
#4
concentracion = 100
ciclos = 0

while concentracion >= 10:
    concentracion = concentracion * 0.8
    ciclos = ciclos + 1
    print("Ciclo:", ciclos)
    print("Concentracion:", concentracion)

print("Necesita", ciclos, "ciclos")
```

```python
#5
opcion = ""

while opcion != "2":
    print("1. Convertir gramos a kilogramos")
    print("2. Salir")

    opcion = input("Seleccione una opcion: ")

    if opcion == "1":
        gramos = float(input("Ingrese los gramos: "))
        kilogramos = gramos / 1000
        print("Kilogramos:", kilogramos)

    elif opcion == "2":
        print("Programa finalizado")

    else:
        print("Opcion invalida")
```

```python
#6
temperatura = 150

while temperatura >= 25:
    print("Temperatura:", temperatura, "°C")
    temperatura = temperatura - 10

print("La sustancia llego a 25 °C o menos")
```

```python
#7
contador = 0
masa_total = 0

while contador < 5:
    masa = float(input("Ingrese una masa positiva: "))

    if masa > 0:
        masa_total = masa_total + masa
        contador = contador + 1
        print("Masa valida")
    else:
        print("La masa debe ser mayor que cero")

print("Masa total:", masa_total)
```

```python
#8
acido = 50
base = 0

while base < 50:
    base = base + 5
    print("Acido:", acido, "mL")
    print("Base agregada:", base, "mL")

print("Se agregaron 50 mL de base")
```

```python
#9
formula = ""

while formula != "H2O":
    formula = input("Ingrese una formula quimica: ")

    if formula != "H2O":
        print("Formula incorrecta")

print("Formula correcta: H2O")
```

```python
#10
nucleos = 800
vidas_medias = 0

while nucleos >= 50:
    nucleos = nucleos / 2
    vidas_medias = vidas_medias + 1
    print("Vida media:", vidas_medias)
    print("Nucleos:", nucleos)

print("Necesita", vidas_medias, "vidas medias")
```
