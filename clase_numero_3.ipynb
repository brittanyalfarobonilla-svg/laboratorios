```python
print("indice de masa corporal")
peso=float(input("digite su peso en kg: "))
altura = float(input("digite su estatura en metros: "))
imc = peso/altura**2
print("su indece de masa corporal es: ", imc)

print("---------------------------------------------------")
if imc < 18.5:
  print("infrapeso")
elif imc < 25:
  print("peso normal")
elif imc < 30:
  print("sobrepeso")
elif imc < 40:
  print("obesidad")
else:
  print("obesidad morbida")
```
