```python
#solicitar dos numeros
numero1 = float(input("ingrese el primer numero  "))
numero2 = float(input("ingrese el segundo numero  "))
# operacion
suma= numero1 + numero2
resta= numero1 - numero2
multiplicacion= numero1 * numero2
#mostrar resultados
print("Los resultados son")
print("El resultado de la suma es:", suma)
print("El resultado de la resta es:", resta)
print("El resultado de la multiplicacion es: ", multiplicacion)
```

```python
# solicitar datos
print("calcular el area y el prerimetro de un rectangulo")
base= float(input("ingrese la base: "))
altura= float(input("ingrese la altura: "))
# operaciones
prerimetro= 2* (base + altura)
area= base * altura
# mostrar resultados
print("Los resultados son")
print("El prerimetro de el rectangulo es: ", prerimetro)
print("El area del rectangulo es : ", area)

```

```python
# solicitar precio
precio= float(input("Ingrese el precio del producto: "))
# operacion del 15%
Descuento= precio * 0.15
precio_final= precio - Descuento
print("descuento del 15%", Descuento)
print("el precio final es: ", precio_final)
```

```python
print("conversion de horas a minutos y segundos")
horas= float(input("ingrese la cantidad de horas: "))
minutos= horas * 60
segundos= horas * 3600
print("resultados")
print("horas: ", horas)
print("minutos: ", minutos)
print("segundos: ", segundos)
```
