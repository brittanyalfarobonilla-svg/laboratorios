```python
import pandas as pd
import matplotlib.pyplot as plt

datos = pd.read_csv("casasboston.csv")

df = datos.rename(columns={
    "ciudad": "CIUDAD",
    "habitaciones": "N_HABITACIONES_MEDIO",
    "valor_mediano": "VALOR_MEDIANO"
})

# Reto: gráfico de pie con etiquetas
df.CIUDAD.value_counts().plot.pie(
    labels=df.CIUDAD.value_counts().index,
    autopct="%1.1f%%"
)

plt.title("Cantidad de casas por ciudad")
plt.ylabel("")
plt.show()
```
