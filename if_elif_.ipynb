```python
#1
ph = float(input("Ingresa el pH: "))

if ph < 7:
    print("La sustancia es ácida.")
elif ph == 7:
    print("La sustancia es neutra.")
else:
    print("La sustancia es básica.")
```

```python
#2
temperatura = float(input("Ingresa la temperatura en °C: "))

if temperatura < 0:
    print("El agua está en estado sólido.")
elif temperatura <= 100:
    print("El agua está en estado líquido.")
else:
    print("El agua está en estado gaseoso.")
```

```python
#3
simbolo = input("Ingresa el símbolo del elemento: ")

if simbolo in ["Fe", "Cu", "Na", "Mg"]:
    print("Es un metal.")
elif simbolo in ["H", "C", "N", "O", "F", "Cl"]:
    print("Es un no metal.")
elif simbolo in ["He", "Ne", "Ar"]:
    print("Es un gas noble.")
else:
    print("Elemento no incluido.")
```

```python
#4
concentracion = float(input("Ingresa la concentración en mol/L: "))

if concentracion < 0.5:
    print("La disolución es diluida.")
elif concentracion <= 2:
    print("La disolución es moderada.")
else:
    print("La disolución es concentrada.")
```

```python
#5
temperatura = float(input("Ingresa la temperatura de la reacción en °C: "))

if temperatura < 50:
    print("Riesgo bajo.")
elif temperatura <= 100:
    print("Riesgo moderado.")
else:
    print("Riesgo alto.")
```

```python
#6
numero = int(input("Ingresa un número atómico entre 1 y 5: "))

if numero == 1:
    print("Hidrógeno (H)")
elif numero == 2:
    print("Helio (He)")
elif numero == 3:
    print("Litio (Li)")
elif numero == 4:
    print("Berilio (Be)")
elif numero == 5:
    print("Boro (B)")
else:
    print("Número atómico no válido.")
```

```python
#7
e1 = float(input("Ingresa la electronegatividad del primer elemento: "))
e2 = float(input("Ingresa la electronegatividad del segundo elemento: "))

if e1 > e2:
    print("El primer elemento tiene mayor electronegatividad.")
elif e2 > e1:
    print("El segundo elemento tiene mayor electronegatividad.")
else:
    print("Ambos tienen la misma electronegatividad.")
```

```python
#8
diferencia = float(input("Ingresa la diferencia de electronegatividad: "))

if diferencia < 0.4:
    print("Enlace covalente no polar.")
elif diferencia < 1.7:
    print("Enlace covalente polar.")
else:
    print("Enlace iónico.")
```

```python
#9
A = float(input("Ingrese la cantidad de A: "))
B = float(input("Ingrese la cantidad de B: "))

if A < B:
    print("El reactivo limitante es A")
elif B < A:
    print("El reactivo limitante es B")
else:
    print("Ningún reactivo está en exceso")
```

```python
#10
rendimiento = float(input("Ingrese el rendimiento porcentual: "))

if rendimiento < 50:
    print("Rendimiento bajo")
elif rendimiento < 80:
    print("Rendimiento moderado")
else:
    print("Rendimiento alto")
```
