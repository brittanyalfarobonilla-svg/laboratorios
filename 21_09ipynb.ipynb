```python
#importar folium para mapas
import folium

#importar MiniMap
from folium.plugins import MiniMap

#cordenadas de heredia
latitud=10.000000
longitud=-84.000000

#crear el mapa
my_map=folium.Map(location=[latitud,longitud], zoom_start=12, titles='OpenStreeMap')
#texto que aparecerá al hacer clic
popuptext='<b>Heredia, Costa Rica</b>'

#agregar marcador
folium.Marker(
    [latitud, longitud],
    popup=popuptext,
    tooltip='Heredia',
    icon=folium.Icon(color='orange', icon='info-sign',prefix='fa')
).add_to(my_map)

#agregar circulo
folium.Circle(
    radius=-1000,
    location=[latitud, longitud],
              color='purple',
              fill=True,
              fill_color='red',
              fill_opacity=0.3,
              tooltip='Zona Heredia').add_to(my_map)

#agregar minimapa
minimap=MiniMap(tile_layer='OpenStreetMap', possition='bottomright')
my_map.add_child(minimap)
#guardar mapa
my_map.save('Heredia.html')

#mostrar el mapa
my_map

```
