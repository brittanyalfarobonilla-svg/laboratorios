```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
#leer archivo csv
df = pd.read_csv ("/content/School-Age-Digital-Connectivity - Primary.csv")

#crear una tabla de contigencia(cuenta las frecuencias de combinaciones de income gruoup y residence urban)
contingency_table = pd.crosstab (df ["Income Group"], df["Region"])
df.head()
#crear mapa de calor usando seaborn. la biblioteca trae la funcion "heatamap",
#annot:este parametro realiza anotaciones en celda del mapa
#cmap permite personalizar el mapa


#crear el mapa de calor
plt.figure(figsize=(10, 7))
sns.heatmap(contingency_table, annot=True, cmap= "PiYG")
plt.title("Contingency Tabla: Income Group vs Residence Urban")
plt.show()
#mostrar grafico
plt.title("Contigency Table: Income Group vs Residence Urban")
plt.show()

```
